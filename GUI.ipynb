```python
import numpy as np
```

```python
def caffeine_at(t, doses, half_life=5):
    return sum(mg * 0.5 ** ((t - td) / half_life) for td, mg in doses if t >= td)
```

```python
def latest_safe_time(bed, doses, new_mg=95, threshold=25, half_life=5):
    room = threshold - caffeine_at(bed, doses, half_life)
    if room <= 0:
        return None
    return bed - half_life * np.log2(new_mg / room)
```

```python
doses = [(8,95)]
```

```python
print(caffeine_at(23, doses))
```

```python
print(latest_safe_time(23, doses))
```
