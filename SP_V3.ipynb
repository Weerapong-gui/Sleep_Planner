# Coffee Planner — ดื่มกาแฟกี่โมง ถึงจะนอนหลับ?
แนวคิดหลัก: ทุก 5 ชั่วโมง คาเฟอีนในร่างกายจะเหลือครึ่งหนึ่ง ค่าครึ่งชีวิต (Half-life)

```python
import matplotlib.pyplot as plt
```

```python
HALF_LIFE = 5

def caffeine_left(dose, hours):
    n = hours / HALF_LIFE
    return dose * 0.5 ** n
```

```python
def show_table(dose, drink_hour, bed_hour):
    for h in range(drink_hour, bed_hour + 1):
        left = caffeine_left(dose, h - drink_hour)
        print(f"{h % 24}:00 caffeine left > {left:.2f} mg")#มีการPrint อยู่แล้วในFuntion
```

เราจะใช้ (proposition) จาก Lec 4-5 


```python
SLEEP_LIMIT = 50
DAILY_LIMIT = 400

def can_sleep(dose, drink_hour, bed_hour):
    left = caffeine_left(dose, bed_hour - drink_hour)
    p = left <= SLEEP_LIMIT
    q = dose <= DAILY_LIMIT
    return p and q 
```

```python
def plot_caffeine(dose, drink_hour, bed_hour):
    hours = list(range(drink_hour, bed_hour + 1))

    amounts = [caffeine_left(dose, h - drink_hour) for h in hours]

    plt.plot(hours, amounts, marker="o", label="Caffeine in body")
    plt.axhline(SLEEP_LIMIT, color="red", linestyle= "--", label="Sleep limit 50 mg")
    plt.axvline(bed_hour, color="gray", linestyle=":", label="Bedtime")

    plt.xlabel("Time (hour)")
    plt.ylabel("Caffeine (mg)")
    plt.title("Caffeine remaining after drinking")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
def main ():
    #input value from user 
    dose = int(input("ปริมาณคาเฟอีนที่กำลังจะดื่ม (mg) :"))
    drink_hour = int(input("ช่วงเวลาดื่ม 0 - 23 :"))
    bed_hour = int(input("ช่วงเวลานอน 0 - 23 :"))

    if bed_hour <= drink_hour:
        bed_hour = bed_hour + 24
    
    #showTable
    show_table(dose, drink_hour, bed_hour)
    plot_caffeine(dose, drink_hour, bed_hour)

    #discut result
    if can_sleep(dose, drink_hour, bed_hour):
        print("นอนได้!")
    else:
        print("นอนไม่ได้")
```

```python
main()
```
